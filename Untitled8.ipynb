```python
import os, io, time, zipfile, cv2
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
from google.colab import files
from tensorflow.keras import Model
from tensorflow.keras.layers import Input, Conv2D, SeparableConv2D, Add, ReLU, Lambda

SIZE, F, EPOCHS, SAMPLES, BATCH = 128, 16, 10, 48, 8

uploaded = files.upload()
name = next(iter(uploaded))
data = uploaded[name]
ext = os.path.splitext(name)[1].lower()

def load_file(data, ext):
    if ext in [".jpg",".jpeg",".png",".bmp",".webp"]:
        img = cv2.imdecode(np.frombuffer(data,np.uint8),cv2.IMREAD_COLOR)
        if img is None: raise ValueError("Invalid image")
        return cv2.cvtColor(img,cv2.COLOR_BGR2RGB)

    if ext == ".pdf":
        try: import fitz
        except:
            !pip -q install pymupdf
            import fitz
        pdf = fitz.open(stream=data,filetype="pdf")
        pix = pdf[0].get_pixmap(matrix=fitz.Matrix(2,2),alpha=False)
        return np.frombuffer(pix.samples,np.uint8).reshape(
            pix.height,pix.width,pix.n
        )[:,:,:3]

    if ext == ".docx":
        with zipfile.ZipFile(io.BytesIO(data)) as z:
            imgs = [x for x in z.namelist()
                    if x.startswith("word/media/") and
                    x.lower().endswith((".png",".jpg",".jpeg",".bmp"))]
            if imgs:
                img = cv2.imdecode(
                    np.frombuffer(z.read(imgs[0]),np.uint8),
                    cv2.IMREAD_COLOR
                )
                return cv2.cvtColor(img,cv2.COLOR_BGR2RGB)

    raise ValueError("Use JPG, JPEG, PNG, PDF or DOCX")

original = cv2.resize(
    load_file(data,ext),
    (SIZE,SIZE),
    interpolation=cv2.INTER_AREA
).astype(np.float32)/255.0

def traditional(img):
    x = (img*255).astype(np.uint8)
    x = cv2.bilateralFilter(x,5,35,35)
    lab = cv2.cvtColor(x,cv2.COLOR_RGB2LAB)
    l,a,b = cv2.split(lab)
    l = cv2.createCLAHE(
        clipLimit=1.4,
        tileGridSize=(8,8)
    ).apply(l)
    x = cv2.cvtColor(
        cv2.merge((l,a,b)),
        cv2.COLOR_LAB2RGB
    )
    blur = cv2.GaussianBlur(x,(0,0),1)
    x = cv2.addWeighted(x,1.12,blur,-0.12,0)
    return np.clip(x/255.0,0,1).astype(np.float32)

def degrade(img):
    x = (img*255).astype(np.uint8)
    scale = np.random.choice([.5,.625,.75])
    small = cv2.resize(
        x,
        (int(SIZE*scale),int(SIZE*scale)),
        interpolation=cv2.INTER_AREA
    )
    x = cv2.resize(
        small,(SIZE,SIZE),
        interpolation=cv2.INTER_LINEAR
    )
    x = cv2.GaussianBlur(
        x,(3,3),np.random.uniform(.4,1)
    )
    x = np.clip(
        x.astype(np.float32)+np.random.normal(
            0,np.random.uniform(3,8),x.shape
        ),
        0,255
    ).astype(np.uint8)
    q = np.random.randint(60,85)
    ok,enc = cv2.imencode(
        ".jpg",x,[int(cv2.IMWRITE_JPEG_QUALITY),q]
    )
    if ok:
        x = cv2.cvtColor(
            cv2.imdecode(enc,cv2.IMREAD_COLOR),
            cv2.COLOR_BGR2RGB
        )
    return x.astype(np.float32)/255.0

train_x = np.array(
    [degrade(original) for _ in range(SAMPLES)],
    dtype=np.float32
)

train_y = np.repeat(
    original[None],SAMPLES,axis=0
).astype(np.float32)

def block(x):
    return ReLU()(
        SeparableConv2D(
            F,3,padding="same",use_bias=False
        )(x)
    )

inp = Input((SIZE,SIZE,3))
x = ReLU()(
    Conv2D(F,3,padding="same",use_bias=False)(inp)
)
x = block(x)
x = block(x)
x = block(x)

residual = Conv2D(3,1,padding="same")(x)
out = Add()([inp,.5*residual])
out = Lambda(lambda x:tf.clip_by_value(x,0,1))(out)

model = Model(inp,out,name="LightweightGameEnhancer")

def loss(y_true,y_pred):
    mse = tf.reduce_mean(tf.square(y_true-y_pred))
    ssim = 1-tf.reduce_mean(
        tf.image.ssim(y_true,y_pred,max_val=1.0)
    )
    return .4*mse+.6*ssim

model.compile(
    optimizer=tf.keras.optimizers.Adam(.001),
    loss=loss
)

print("Parameters:",model.count_params())

start = time.perf_counter()

model.fit(
    train_x,train_y,
    batch_size=BATCH,
    epochs=EPOCHS,
    validation_split=.2,
    verbose=1
)

training_time = time.perf_counter()-start

model.save("/content/lightweight_game_enhancer.keras")

start = time.perf_counter()

cnn = model(
    original[None],
    training=False
).numpy()[0]

inference_time = (time.perf_counter()-start)*1000
cnn = np.clip(cnn,0,1)

trad = traditional(train_x[0])

def metrics(a,b):
    a,b = map(
        lambda x:tf.convert_to_tensor(x[None],tf.float32),
        [a,b]
    )
    return (
        tf.image.psnr(a,b,max_val=1).numpy()[0],
        tf.image.ssim(a,b,max_val=1).numpy()[0]
    )

p_deg,s_deg = metrics(original,train_x[0])
p_trad,s_trad = metrics(original,trad)
p_cnn,s_cnn = metrics(original,cnn)
p_ct,s_ct = metrics(trad,cnn)

print("\n======================================")
print("MODEL PERFORMANCE")
print("======================================")
print(f"Parameters     : {model.count_params():,}")
print(f"Inference time : {inference_time:.3f} ms")
print(f"Training time  : {training_time:.2f} sec")

print("\n======================================")
print("QUALITY COMPARISON")
print("======================================")
print(f"Degraded vs Original")
print(f"PSNR : {p_deg:.4f} dB")
print(f"SSIM : {s_deg:.4f}")

print(f"\nTraditional vs Original")
print(f"PSNR : {p_trad:.4f} dB")
print(f"SSIM : {s_trad:.4f}")

print(f"\nCNN Enhanced vs Original")
print(f"PSNR : {p_cnn:.4f} dB")
print(f"SSIM : {s_cnn:.4f}")

print(f"\nCNN Enhanced vs Traditional")
print(f"PSNR : {p_ct:.4f} dB")
print(f"SSIM : {s_ct:.4f}")

print("\n======================================")
print("IMPROVEMENT")
print("======================================")
print(f"PSNR Gain : {p_cnn-p_deg:+.4f} dB")
print(f"SSIM Gain : {s_cnn-s_deg:+.4f}")

plt.figure(figsize=(12,7))

for i,(img,title) in enumerate([
    (original,"Original"),
    (train_x[0],"Degraded Input"),
    (trad,"Traditional"),
    (cnn,"CNN Enhanced")
]):
    plt.subplot(2,2,i+1)
    plt.imshow(img)
    plt.title(title)
    plt.axis("off")

plt.tight_layout()
plt.show()

cv2.imwrite(
    "/content/enhanced_output.png",
    cv2.cvtColor(
        (cnn*255).astype(np.uint8),
        cv2.COLOR_RGB2BGR
    )
)

print("\nSaved: /content/enhanced_output.png")
```
